### Story Pitch

India’s wholesale inflation rose to 9.9% in August 2026, but the headline masks a split within the basket. Fuel and power prices rose 22.9% year-on-year, led by mineral oils and crude, while electricity prices fell. Yet manufactured products, with a 63.1% weight, contributed the most to the overall rise. The story will examine how a sharp energy shock and broad-based manufacturing inflation combined to push WPI higher across the economy.



```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the WPI dataset
file_path = r"C:\Users\abhis\Downloads\data acj\wpi_monthly_index_202609.xlsx"

df = pd.read_excel(r"C:\Users\abhis\Downloads\data acj\wpi_monthly_index_202609.xlsx")


# Convert monthly WPI columns to numeric
month_cols = df.columns[4:]

for col in month_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")


# Select the three Major Groups
major = df[df["Level"] == "Major Group"].copy()


# Give the groups shorter names
def clean_name(name):

    if "PRIMARY" in str(name).upper():
        return "Primary articles"

    elif "FUEL" in str(name).upper():
        return "Fuel & power"

    elif "MANUFACTURED" in str(name).upper():
        return "Manufactured products"


major["Major group"] = major["Commodity Name"].apply(clean_name)


# Calculate August 2026 year-on-year inflation
major["YoY inflation"] = (
    major["Aug-26"] /
    major["Aug-25"] -
    1
) * 100


# Calculate weighted price pressure
major["Weighted pressure"] = (
    major["Commodity Weight"] *
    major["YoY inflation"] /
    100
)


# Put groups in the order we want
order = [
    "Manufactured products",
    "Fuel & power",
    "Primary articles"
]

major = (
    major
    .set_index("Major group")
    .loc[order]
    .reset_index()
)


# Create the chart
fig, ax = plt.subplots(figsize=(11, 6))

bars = ax.barh(
    major["Major group"],
    major["Weighted pressure"]
)

ax.invert_yaxis()


# Add explanatory labels
for bar, (_, row) in zip(bars, major.iterrows()):

    text = (
        f"{row['Weighted pressure']:.2f} pp"
        f"   |   {row['Commodity Weight']:.2f}% basket"
        f"   |   +{row['YoY inflation']:.2f}% YoY"
    )

    ax.text(
        row["Weighted pressure"] + 0.08,
        bar.get_y() + bar.get_height() / 2,
        text,
        va="center",
        fontsize=10
    )


# Title
ax.set_title(
    "Manufacturing carries the largest weighted price pressure",
    fontsize=15,
    pad=18
)


# Subtitle
ax.text(
    0.5,
    1.02,
    "August 2026 | Bar length = WPI weight × year-on-year price increase",
    transform=ax.transAxes,
    ha="center",
    fontsize=10
)


# Axis titles
ax.set_xlabel(
    "Approximate weighted price pressure (percentage points)"
)

ax.set_ylabel("Major WPI group")


# Clean presentation
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.grid(
    axis="x",
    alpha=0.2
)

plt.tight_layout()

plt.savefig(
    "Chart_2_WPI_weighted_pressure.png",
    dpi=220,
    bbox_inches="tight"
)

plt.show()
plt.savefig(
    r"C:\Users\abhis\Downloads\data acj\Chart_2_WPI_weighted_pressure.png",
    dpi=220,
    bbox_inches="tight"
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the WPI dataset
file_path = r"C:\Users\abhis\Downloads\data acj\wpi_monthly_index_202609.xlsx"

df = pd.read_excel(r"C:\Users\abhis\Downloads\data acj\wpi_monthly_index_202609.xlsx")


# Convert monthly WPI columns to numeric
month_cols = df.columns[4:]

for col in month_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")


# Select Group-level observations
groups = df[df["Level"] == "Group"].copy()


# Keep manufacturing groups
manufacturing = groups[
    groups["Commodity Name"]
    .astype(str)
    .str.contains(
        "MANUFACTURE",
        case=False,
        na=False
    )
].copy()


# Convert long names into short names
def short_name(name):

    name = str(name).upper()

    if "TOBACCO" in name:
        return "Tobacco"

    elif "TEXTILES" in name:
        return "Textiles"

    elif "CHEMICALS" in name:
        return "Chemicals"

    elif "RUBBER" in name and "PLASTIC" in name:
        return "Rubber & plastics"

    elif "BASIC METALS" in name:
        return "Basic metals"

    elif "ELECTRICAL EQUIPMENT" in name:
        return "Electrical equipment"

    elif "OTHER MANUFACTURING" in name:
        return "Other manufacturing"


manufacturing["Group"] = (
    manufacturing["Commodity Name"]
    .apply(short_name)
)


# Calculate August 2026 inflation
manufacturing["Aug-26 YoY"] = (
    manufacturing["Aug-26"] /
    manufacturing["Aug-25"] -
    1
) * 100


# Keep groups with August inflation of 10% or more
selected = manufacturing[
    manufacturing["Aug-26 YoY"] >= 10
].copy()


# Months we want to display
months = [
    "Apr-26",
    "May-26",
    "Jun-26",
    "Jul-26",
    "Aug-26"
]


# Create the heatmap data
heatmap_values = []

for _, row in selected.iterrows():

    row_values = []

    for month in months:

        current_date = pd.to_datetime(
            month,
            format="%b-%y"
        )

        previous_date = (
            current_date -
            pd.DateOffset(years=1)
        )

        previous_month = previous_date.strftime(
            "%b-%y"
        )

        yoy = (
            row[month] /
            row[previous_month] -
            1
        ) * 100

        row_values.append(yoy)

    heatmap_values.append(row_values)


# Create dataframe
heatmap = pd.DataFrame(
    heatmap_values,
    index=selected["Group"],
    columns=["Apr", "May", "Jun", "Jul", "Aug"]
)


# Sort according to August inflation
heatmap = heatmap.sort_values(
    "Aug",
    ascending=True
)


# Create the heatmap
fig, ax = plt.subplots(
    figsize=(11, 7)
)

image = ax.imshow(
    heatmap.values,
    aspect="auto"
)


# X-axis
ax.set_xticks(
    range(len(heatmap.columns))
)

ax.set_xticklabels(
    heatmap.columns
)


# Y-axis
ax.set_yticks(
    range(len(heatmap.index))
)

ax.set_yticklabels(
    heatmap.index
)


# Put the actual inflation rate inside every cell
for i in range(heatmap.shape[0]):

    for j in range(heatmap.shape[1]):

        value = heatmap.iloc[i, j]

        ax.text(
            j,
            i,
            f"{value:.1f}%",
            ha="center",
            va="center",
            fontsize=10
        )


# Title
ax.set_title(
    "Manufacturing price inflation stayed elevated through the summer",
    fontsize=15,
    pad=18
)


# Subtitle
ax.text(
    0.5,
    1.02,
    "Year-on-year WPI inflation, April–August 2026",
    transform=ax.transAxes,
    ha="center",
    fontsize=10
)


# Axis titles
ax.set_xlabel("Month")
ax.set_ylabel("Manufacturing group")


# Colour scale
cbar = fig.colorbar(
    image,
    ax=ax
)

cbar.set_label(
    "Year-on-year inflation (%)"
)


# Bottom explanation
ax.text(
    0.5,
    -0.14,
    "Seven manufacturing groups were above 10% in August; together they represent 26.30% of the WPI basket.",
    transform=ax.transAxes,
    ha="center",
    fontsize=10
)


plt.tight_layout()

plt.savefig(
    "Chart_3_WPI_manufacturing_persistence.png",
    dpi=220,
    bbox_inches="tight"
)

plt.show()
plt.savefig(
    r"C:\Users\abhis\Downloads\data acj\Chart_3_WPI_manufacturing_persistence.png",
    dpi=220,
    bbox_inches="tight"
)
```
